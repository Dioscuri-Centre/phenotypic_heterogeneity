```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
conditions=['control','dmso','10 $\mu$M','20 $\mu$M','50 $\mu$M','100 $\mu$M','200 $\mu$M','500 $\mu$M']
```

```python
colonies=[np.array([241,231,183,140,54,48,33,0]),#old clonogenic assay1 with u87mg WT cells
    np.array([185,173,134,109,46,34,12,0]),#old clonogenic assay2 with u87mg WT cells
    np.array([130,114,84,64,25,15,8,0]),#new clonogenic assay1 with u87mg WT cells
    np.array([111,110,71,55,23,18,6,0]),#new clonogenic assay2 with u87mg WT cells
    np.array([126,111,81,45,23,18,3,0])]#new clonogenic assay3 with u87mg WT cells
```

```python
for i in range(0,5):
    plt.plot(conditions,colonies[i]/colonies[i][0])
#plt.errorbar(conditions,colonies1)
plt.xlabel('conditons')
plt.ylabel('no. of colonies')
plt.show()
```

```python
plt.rcParams['font.size'] = '10'
plt.rcParams["figure.figsize"] = (6,1.5)
plt.figure(constrained_layout=True)
ratio=np.mean([c/c[0] for c in colonies],axis=0)
err=np.std([c/c[0] for c in colonies],axis=0)
plt.bar(conditions, ratio,yerr=err,capsize=5)
plt.ylabel("$n/n_{DMSO}$")
plt.savefig('../plots/clonogenic_assay.svg',dpi=300)
plt.show()
```
